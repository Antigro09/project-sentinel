# X65A finite-theory reproduction

This notebook launches the checked scripts. Outputs are finite toy-model evidence, not an X65A gate pass. Run it only inside the project math-research virtual environment.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd()
scripts = [
    'x65a_symbolic_checks.py',
    'x65a_exact_checks.py',
    'x65a_property_checks.py',
    'x65a_resource_checks.py',
]
for script in scripts:
    subprocess.run([sys.executable, str(root / script)], check=True)
```

```python
import json
for result in sorted((root / 'results').glob('*.json')):
    payload = json.loads(result.read_text())
    print(result.name, '->', payload['status'])
```
